```python
f = h5py.File('test_schema_view.h5', 'w')
```

Creates folder and subfolders

```python
runs = f.create_group("Runs")
micro_data = runs.create_group("micro_data")
macro_data = runs.create_group("macro_data")
```

Instantiates several 50000 x 1 arrays for microscale data

```python
first_pli = micro_data.create_dataset("first_pli", (50000 , 1), dtype='d')
last_tpa = micro_data.create_dataset("last_tpa", (50000 , 1), dtype='i')
lys_complete = micro_data.create_dataset("lys_complete", (50000 , 1), dtype='i1')
lysis = micro_data.create_dataset("lysis", (50000 , 1), dtype='d')
pli = micro_data.create_dataset("pli", (50000 , 1), dtype='i')
tpa_time = micro_data.create_dataset("tpa_time", (50000 , 1), dtype='d')
tPAPLiunbd = micro_data.create_dataset("tPAPLiunbd", (50000 , 1), dtype='i1')
tPAunbind = micro_data.create_dataset("tPAunbind", (50000 , 1), dtype='i1')
```

Instantiations of more data sets for macroscale data

```python
fiber_degrade_time = macro_data.create_dataset("fiber_degrade_time", (1 , 3) , dtype='f8', maxshape = (None, 3))
tpa_bind_events = macro_data.create_dataset("tpa_bind_events", (1 , 10) , dtype='f8' , maxshape = (None, 10))
tpa_location_snapshot = macro_data.create_dataset("tpa_location_snapshot", (e.macro_params.total_molecules , 1) , dtype='i' , maxshape = (e.macro_params.total_molecules, None)) 
tpa_transit_time = macro_data.create_dataset("tpa_transit_time", (e.macro_params.total_molecules , 1) , dtype='f8')
snapshot_time = macro_data.create_dataset("snapshot_time", (1 , 1) , dtype='f8' , maxshape = (None, 1))
```
